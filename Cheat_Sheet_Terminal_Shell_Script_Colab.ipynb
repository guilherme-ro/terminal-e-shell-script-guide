## 1. Navegação de Diretórios

```python
# Mostrar diretório atual
!pwd

# Listar arquivos (simples)
!ls

# Listar com detalhes + arquivos ocultos
!ls -la

# Listar de forma legível (human-readable)
!ls -lh

# Entrar em um diretório
!cd /tmp && pwd

# Voltar um nível
!cd .. && pwd

# Ir para o home do usuário
!cd ~ && pwd

# Criar e entrar em um diretório de uma vez
!mkdir -p ~/meu_teste && cd ~/meu_teste && pwd
```

## 2. Manipulação de Arquivos e Diretórios

```python
# Criar arquivo vazio
!touch arquivo_teste.txt

# Criar diretório
!mkdir pasta_teste

# Criar diretórios aninhados
!mkdir -p pasta1/pasta2/pasta3

# Copiar arquivo
!cp arquivo_teste.txt copia.txt

# Copiar diretório recursivamente
!cp -r pasta_teste pasta_copia

# Mover / Renomear
!mv copia.txt renomeado.txt

# Remover arquivo
!rm renomeado.txt

# Remover diretório vazio
!rmdir pasta_teste 2>/dev/null || true

# Remover diretório com conteúdo (CUIDADO!)
!rm -rf pasta_copia

# Ver conteúdo de um arquivo
!echo "Olá mundo" > exemplo.txt
!cat exemplo.txt

# Ver primeiras linhas
!head -n 5 /etc/passwd

# Ver últimas linhas
!tail -n 5 /etc/passwd

# Contar linhas, palavras e bytes
!wc -l /etc/passwd
!wc -w /etc/passwd
!wc -c /etc/passwd
```

## 3. Visualização e Edição Rápida de Texto

```python
# Criar um arquivo de exemplo
%%writefile dados.txt
linha 1: banana
linha 2: maçã
linha 3: banana
linha 4: laranja
linha 5: banana
```

```python
# grep - buscar padrões
!grep banana dados.txt

# grep com número da linha
!grep -n banana dados.txt

# grep invertido (linhas que NÃO contêm)
!grep -v banana dados.txt

# grep case-insensitive
!grep -i BANANA dados.txt

# Contar ocorrências
!grep -c banana dados.txt

# cut - extrair colunas
!echo "nome:idade:cidade" | cut -d: -f1
!echo "nome:idade:cidade" | cut -d: -f2

# sort e uniq
!sort dados.txt
!sort dados.txt | uniq
!sort dados.txt | uniq -c   # conta ocorrências

# sed - substituição simples
!sed 's/banana/abacaxi/g' dados.txt

# awk - processar colunas
!awk '{print $2}' dados.txt
!awk -F: '{print $1}' /etc/passwd | head -5
```

## 4. Permissões e Propriedade

```python
# Ver permissões
!ls -l dados.txt

# Dar permissão de execução
!chmod +x dados.txt
!ls -l dados.txt

# Remover permissão de escrita para grupo e outros
!chmod go-w dados.txt
!ls -l dados.txt

# Permissão numérica (ex: 755)
!chmod 755 dados.txt
!ls -l dados.txt

# Ver dono e grupo
!ls -l dados.txt

# Mudar dono (pode precisar de sudo - no Colab geralmente não funciona)
# !chown root dados.txt
```

## 5. Processos e Sistema

```python
# Ver processos em execução
!ps aux | head -10

# Ver processos de forma interativa (melhor usar no terminal real)
# !top

# Ver uso de memória
!free -h

# Ver uso de disco
!df -h

# Ver uso de um diretório específico
!du -sh /content 2>/dev/null || du -sh .

# Ver informações do sistema
!uname -a
!cat /etc/os-release | head -5

# Ver data e hora
!date

# Ver quem está logado
!whoami
!id
```

## 6. Redirecionamento e Pipes

```python
# Redirecionar saída para arquivo (sobrescreve)
!echo "primeira linha" > saida.txt
!cat saida.txt

# Acrescentar ao arquivo
!echo "segunda linha" >> saida.txt
!cat saida.txt

# Redirecionar erro
!ls /diretorio_inexistente 2> erro.txt
!cat erro.txt

# Pipe (|) - passar saída de um comando para outro
!ls -la | grep ".txt"

# Combinar vários pipes
!cat /etc/passwd | grep bash | cut -d: -f1 | head -5

# tee - mostrar e salvar ao mesmo tempo
!echo "teste com tee" | tee arquivo_tee.txt
!cat arquivo_tee.txt
```

## 7. Compactação e Arquivos

```python
# Criar arquivo tar.gz
!tar -czf meus_arquivos.tar.gz dados.txt saida.txt
!ls -lh *.tar.gz

# Listar conteúdo do tar
!tar -tzf meus_arquivos.tar.gz

# Extrair
!mkdir -p extracao && tar -xzf meus_arquivos.tar.gz -C extracao
!ls extracao

# zip / unzip (se disponível)
!zip arquivo.zip dados.txt saida.txt 2>/dev/null || echo "zip não instalado"
!unzip -l arquivo.zip 2>/dev/null || true
```

## 9. Shell Script — Fundamentos

### 9.1 Criando e executando um script

```python
%%writefile ola.sh
#!/bin/bash
echo "Olá, $USER!"
echo "Data de hoje: $(date)"
echo "Diretório atual: $(pwd)"
echo "Argumentos recebidos: $@"
echo "Número de argumentos: $#"
```

```python
# Dar permissão de execução e rodar
!chmod +x ola.sh
!./ola.sh argumento1 argumento2

# Ou executar com bash diretamente
!bash ola.sh teste
```

### 9.2 Variáveis

```python
%%writefile variaveis.sh
#!/bin/bash

# Variáveis simples
NOME="Grok"
IDADE=1

echo "Nome: $NOME"
echo "Idade: $IDADE"

# Concatenação
echo "Olá, meu nome é ${NOME} e tenho ${IDADE} ano(s)."

# Variáveis de ambiente
echo "Usuário: $USER"
echo "Home: $HOME"
echo "PATH: $PATH" | head -c 80
echo "..."

# Ler entrada do usuário (no Colab funciona limitado)
# read -p "Digite seu nome: " NOME_USUARIO
# echo "Você digitou: $NOME_USUARIO"

# Aritmética
A=10
B=3
echo "Soma: $((A + B))"
echo "Multiplicação: $((A * B))"
echo "Divisão: $((A / B))"
```

```python
!bash variaveis.sh
```

### 9.3 Condicionais (if / else)

```python
%%writefile condicionais.sh
#!/bin/bash

IDADE=18

if [ $IDADE -ge 18 ]; then
    echo "Maior de idade"
else
    echo "Menor de idade"
fi

# Comparações de strings
NOME="Grok"
if [ "$NOME" = "Grok" ]; then
    echo "Nome correto!"
fi

# Verificar se arquivo existe
if [ -f dados.txt ]; then
    echo "Arquivo dados.txt existe"
else
    echo "Arquivo não encontrado"
fi

# Verificar se diretório existe
if [ -d /tmp ]; then
    echo "/tmp existe"
fi

# Operadores numéricos:
# -eq  igual
# -ne  diferente
# -gt  maior que
# -ge  maior ou igual
# -lt  menor que
# -le  menor ou igual

# Operadores de arquivo:
# -f   arquivo regular
# -d   diretório
# -e   existe
# -r   legível
# -w   gravável
# -x   executável
```

```python
!bash condicionais.sh
```

### 9.4 Loops

```python
%%writefile loops.sh
#!/bin/bash

echo "=== for clássico ==="
for i in 1 2 3 4 5; do
    echo "Número: $i"
done

echo ""
echo "=== for com sequência ==="
for i in {1..5}; do
    echo "Contagem: $i"
done

echo ""
echo "=== for em arquivos ==="
for arquivo in *.txt; do
    echo "Arquivo encontrado: $arquivo"
done

echo ""
echo "=== while ==="
contador=1
while [ $contador -le 3 ]; do
    echo "While: $contador"
    contador=$((contador + 1))
done

echo ""
echo "=== until ==="
contador=1
until [ $contador -gt 3 ]; do
    echo "Until: $contador"
    contador=$((contador + 1))
done
```

```python
!bash loops.sh
```

### 9.5 Funções

```python
%%writefile funcoes.sh
#!/bin/bash

# Definir função
saudacao() {
    local nome=$1
    echo "Olá, $nome! Bem-vindo ao Shell Script."
}

# Função com retorno
soma() {
    local a=$1
    local b=$2
    echo $((a + b))
}

# Chamar funções
saudacao "Grok"
saudacao "Usuário do Colab"

resultado=$(soma 10 25)
echo "Resultado da soma: $resultado"
```

```python
!bash funcoes.sh
```

### 9.6 Arrays

```python
%%writefile arrays.sh
#!/bin/bash

frutas=("banana" "maçã" "laranja" "uva")

echo "Primeira fruta: ${frutas[0]}"
echo "Todas as frutas: ${frutas[@]}"
echo "Quantidade: ${#frutas[@]}"

echo ""
echo "Listando com índice:"
for i in "${!frutas[@]}"; do
    echo "  [$i] = ${frutas[$i]}"
done
```

```python
!bash arrays.sh
```

## 10. Exemplos Práticos de Scripts Úteis

```python
%%writefile backup_simples.sh
#!/bin/bash
# Script de backup simples

ORIGEM="."
DESTINO="backup_$(date +%Y%m%d_%H%M%S)"

echo "Criando backup em: $DESTINO"
mkdir -p "$DESTINO"

# Copia apenas arquivos .txt e .sh
cp *.txt *.sh "$DESTINO/" 2>/dev/null

echo "Backup concluído!"
ls -la "$DESTINO"
```

```python
!bash backup_simples.sh
```

```python
%%writefile monitor_disco.sh
#!/bin/bash
# Monitora uso de disco

LIMITE=80
USO=$(df / | tail -1 | awk '{print $5}' | tr -d '%')

echo "Uso atual do disco raiz: ${USO}%"

if [ "$USO" -gt "$LIMITE" ]; then
    echo "⚠️  ALERTA: Disco acima de ${LIMITE}%!"
else
    echo "✅ Disco OK"
fi
```

```python
!bash monitor_disco.sh
```
